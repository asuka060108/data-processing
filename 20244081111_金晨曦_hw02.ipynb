## 一、字符串的乐趣

### 1. `is_palindrome`

回文是指从前往后读和从后往前读都一样的词或短语。判断时忽略空格、标点和大小写，输入只包含英文字母、数字 0–9 和空格。

```python
def is_palindrome(s):
    """判断 s 是否为回文：忽略大小写、空格和标点。"""
    cleaned = [ch.lower() for ch in s if ch.isalnum()]
    return cleaned == cleaned[::-1]
```

```python
print("is_palindrome:")
for s in ["level", "kayak", "pop", "T A C O cat", "rats live on no evil star",
          "Was it a car or a cat I saw?", "hello"]:
    print(f"  {s!r:42} -> {is_palindrome(s)}")
```

**说明：** 先用 `ch.isalnum()` 过滤掉空格和标点，再统一转成小写，最后与反转后的列表比较。`isalnum()` 对英文字母和数字返回 `True`，因此题目中的 `"T A C O cat"` 会被整理为 `"tacocat"`，`"Was it a car or a cat I saw?"` 会被整理为 `"wasitacaroracatisaw"`。空字符串和单个字符也会返回 `True`。

### 2. `is_abecedarian`

如果一个词的字母从左到右按字母顺序排列，就称它是“字母顺序的”，允许字母重复。判断时忽略空格。

```python
def is_abecedarian(s):
    """判断 s 的字母是否按字母表顺序排列（忽略空格，允许重复）。"""
    previous = None
    for ch in s:
        if ch == " ":
            continue
        ch = ch.lower()
        if previous is not None and ch < previous:
            return False
        previous = ch
    return True
```

```python
print("is_abecedarian:")
for s in ["adder", "beet", "abcd efgh xyz", "dog", "cat", "", "a", "aa"]:
    print(f"  {s!r:18} -> {is_abecedarian(s)}")
```

**说明：** 只遍历一次字符串，跳过空格，用 `previous` 记录上一个字母；只要当前字母比上一个字母小，就立即返回 `False`，否则继续。允许相等（重复字母），因此 `"beet"`、`"aa"` 都返回 `True`。空字符串和单字母字符串没有逆序对，返回 `True`。

### 3. `remove_vowels`

去掉字符串中的所有元音 `a`、`e`、`i`、`o`、`u`。

```python
def remove_vowels(s):
    """返回去掉元音 a、e、i、o、u 后的字符串。"""
    vowels = set("aeiouAEIOU")
    return "".join(ch for ch in s if ch not in vowels)
```

```python
print("remove_vowels:")
for s in ["cat", "audio", "aeiou", "Hello World", ""]:
    print(f"  {s!r:15} -> {remove_vowels(s)!r}")
```

**说明：** 用集合保存元音（同时包含大小写），再用生成器表达式保留非元音字符并拼接。`"cat"` 去掉 `a` 得 `"ct"`，`"audio"` 去掉 `a`、`u`、`i`、`o` 得 `"d"`，全部由元音组成的字符串返回空字符串。

## 二、列表的乐趣

### 1. `list_reverse`

接受一个列表，返回反转后的新列表；如果输入不是列表，引发 `TypeError`。

```python
def list_reverse(lst):
    """返回列表的反转副本；输入不是 list 时抛出 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError("list_reverse() 的参数必须是 list")
    return lst[::-1]
```

```python
print("list_reverse:")
print("  [1, 2, 3] ->", list_reverse([1, 2, 3]))

original = [1, 2, 3]
reversed_copy = list_reverse(original)
print("  原列表是否被修改:", original, "（未修改）")
print("  返回值是否为新列表:", reversed_copy is not original)

try:
    list_reverse((1, 2, 3))
except TypeError as exc:
    print("  传入元组时抛出 TypeError:", exc)
```

**说明：** `lst[::-1]` 返回一个新列表，原列表保持不变，符合“返回反转后的列表”的要求。先用 `isinstance(lst, list)` 检查类型，元组、字符串等输入会抛出带有说明信息的 `TypeError`。

### 2. `is_sorted`

接受一个序列 `seq`，如果按升序排列返回 `True`，否则返回 `False`；只遍历序列一次。

```python
def is_sorted(seq):
    """判断序列是否按升序排列；只遍历一次。"""
    it = iter(seq)
    try:
        previous = next(it)
    except StopIteration:
        return True
    for item in it:
        if item < previous:
            return False
        previous = item
    return True
```

```python
print("is_sorted:")
for seq in [[1, 2, 3], [1, 3, 2], [], [5], [1, 1, 2], "abc"]:
    print(f"  {seq!r:15} -> {is_sorted(seq)}")
```

**说明：** 用迭代器从头到尾扫描一次，只保留前一个元素；发现 `item < previous` 就提前返回 `False`。空序列和单元素序列没有逆序对，返回 `True`；相等元素不算逆序，所以 `[1, 1, 2]` 返回 `True`。题目只要求“只遍历一次”，没有要求检查类型，因此字符串、元组等只要元素可比较也能使用。

## 三、作为向量的元组

用长度为 `d` 的元组表示 `d` 维向量，实现标量乘法、内积、矩阵合法性检查和矩阵乘向量。

### 1. `vec_scalar_mult`

接受元组 `t` 和标量 `s`，返回每个元素乘以 `s` 的新元组；类型不合适时抛出 `TypeError`。

```python
def _is_number(value):
    """判断 value 是否为整数或浮点数。"""
    return isinstance(value, (int, float))


def vec_scalar_mult(t, s):
    """返回元组 t 与标量 s 的数乘结果。"""
    if not isinstance(t, tuple):
        raise TypeError("t 必须是元组")
    if not _is_number(s):
        raise TypeError("s 必须是整数或浮点数")
    if not all(_is_number(x) for x in t):
        raise TypeError("t 的元素必须都是整数或浮点数")
    return tuple(x * s for x in t)
```

```python
print("vec_scalar_mult:")
print("  ((1, 2, 3), 2.5) ->", vec_scalar_mult((1, 2, 3), 2.5))
print("  ((1, 2), 3)      ->", vec_scalar_mult((1, 2), 3))

for args in [((1, 2), "3"), ([1, 2], 3), ((1, "a"), 2)]:
    try:
        vec_scalar_mult(*args)
    except TypeError as exc:
        print(f"  {args!r} -> TypeError: {exc}")
```

**说明：** 依次检查 `t` 是否为元组、`s` 是否为数值、`t` 的每个元素是否为数值，任何一项不满足都抛出 `TypeError`；整数标量和整数元素都是有效输入。返回新元组，不修改原元组。

### 2. `vec_inner_product`

接受两个元组，返回内积（浮点数）；任一输入不是元组抛 `TypeError`，长度不一致抛 `ValueError`。

```python
def vec_inner_product(t1, t2):
    """返回两个向量元组的内积（float）。"""
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须一致")
    if not all(_is_number(x) for x in t1) or not all(_is_number(x) for x in t2):
        raise TypeError("向量的元素必须都是整数或浮点数")
    return float(sum(a * b for a, b in zip(t1, t2)))
```

```python
print("vec_inner_product:")
print("  ((1, 2, 3), (4, 5, 6)) ->", vec_inner_product((1, 2, 3), (4, 5, 6)))
print("  ((), ())               ->", vec_inner_product((), ()))

for args in [([1, 2], (1, 2)), ((1, 2), (1, 2, 3)), ((1, "a"), (1, 2))]:
    try:
        vec_inner_product(*args)
    except (TypeError, ValueError) as exc:
        print(f"  {args!r} -> {type(exc).__name__}: {exc}")
```

**说明：** 内积为 `sum(x_j * y_j)`，这里用 `zip` 逐对相乘再求和，最后转成 `float`。类型错误优先检查，长度不一致时抛 `ValueError`；两个空元组的长度为 0，内积按定义为 `0.0`。

### 3. `check_valid_mx`

用“元组的元组”表示 `m × n` 矩阵，判断参数是否为有效矩阵：外层每个元素都是元组、所有内层元组等长、每个元素都是整数或浮点数。

```python
def check_valid_mx(mx):
    """判断 mx 是否为有效矩阵（元组的元组，各行等长，元素为数值）。"""
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if len(mx) == 0:
        return True
    n = len(mx[0])
    if any(len(row) != n for row in mx):
        return False
    return all(_is_number(x) for row in mx for x in row)
```

```python
print("check_valid_mx:")
for mx in [((1, 2), (3, 4)), ((1, 2, 3),), ((1, 2), (3,)), ([1, 2],), ((1, "a"),), ()]:
    print(f"  {mx!r:22} -> {check_valid_mx(mx)}")
```

**说明：** 先检查外层是否为元组，再检查每个元素是否为元组；然后以第一行的长度为基准检查所有行是否等长，最后检查所有元素是否为数值。空元组 `()` 视为合法的 0 行矩阵（没有反例），返回 `True`；`([1, 2],)` 的内层是列表，`((1, "a"),)` 含有字符串，都会返回 `False`。

### 4. `mx_vec_mult`

接受矩阵（元组的元组）和向量（元组），返回矩阵乘向量得到的数值元组；用 `check_valid_mx` 检查矩阵，不合理输入抛出适当异常。

```python
def mx_vec_mult(mx, v):
    """返回矩阵 mx 与向量 v 相乘的结果元组。"""
    if not check_valid_mx(mx):
        raise TypeError("mx 必须是有效矩阵：元组的元组，各行等长，元素为数值")
    if not isinstance(v, tuple):
        raise TypeError("v 必须是元组")
    if not all(_is_number(x) for x in v):
        raise TypeError("v 的元素必须都是整数或浮点数")
    if len(mx) == 0:
        if len(v) != 0:
            raise ValueError("空矩阵只能乘以空向量")
        return ()
    if len(v) != len(mx[0]):
        raise ValueError("向量长度必须等于矩阵的列数")
    return tuple(sum(row[j] * v[j] for j in range(len(v))) for row in mx)
```

```python
print("mx_vec_mult:")
print("  (((1, 2), (3, 4)), (5, 6)) ->", mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
print("  ((), ())                   ->", mx_vec_mult((), ()))

for mx, v in [(((1, 2), (3, 4)), (5,)),
              (((1, 2), (3,)), (5, 6)),
              (([1, 2],), (5, 6))]:
    try:
        mx_vec_mult(mx, v)
    except (TypeError, ValueError) as exc:
        print(f"  {type(exc).__name__}: {exc}")
```

**说明：** 先验证矩阵合法，再验证向量是元组、元素为数值，且长度等于矩阵列数；结果第 `i` 个分量是矩阵第 `i` 行与向量的内积。对 `((1,2),(3,4)) @ (5,6)`，结果为 `(1×5+2×6, 3×5+4×6) = (17, 39)`。矩阵不合法、向量不是元组或元素非数值时抛 `TypeError`；长度不匹配时抛 `ValueError`。

## 四、更多向量的乐趣：`Vector` 类

### 1. 构造与属性

定义 `Vector` 类：每个向量有维度（非负整数）和元素列表/元组。只提供维度时默认生成零向量；维度类型错误或为负、维度与元素个数不一致时抛出异常。

```python
class Vector:
    """用维度与元素表示的向量。"""

    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        if not all(_is_number(x) for x in elements):
            raise TypeError("元素必须都是整数或浮点数")
        self.dim = dim
        self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    def _require_same_dim(self, other):
        """比较/运算前的类型与维度检查。"""
        if not isinstance(other, Vector):
            raise TypeError("只能与另一个 Vector 比较")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度不同，不能比较")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("两个向量的维度不同，不能比较")
        return self.elements == other.elements

    def __lt__(self, other):
        self._require_same_dim(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._require_same_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._require_same_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._require_same_dim(other)
        return self.elements >= other.elements

    def dot(self, other):
        """返回两个 Vector 的内积。"""
        if not isinstance(other, Vector):
            raise TypeError("dot() 的参数必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度不同，不能求内积")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    def __mul__(self, other):
        """标量乘法或两个 Vector 的逐元素乘法。"""
        if _is_number(other):
            return Vector(self.dim, [x * other for x in self.elements])
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量的维度不同，不能相乘")
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        return NotImplemented

    def __rmul__(self, other):
        if _is_number(other):
            return self * other
        return NotImplemented
```

```python
print("构造 Vector:")
print("  Vector(3)             ->", Vector(3))
print("  Vector(2, (2, 4))     ->", Vector(2, (2, 4)))
print("  Vector(0)             ->", Vector(0))

for args in [(-1, [1]), (2, [1]), (2.5, [1, 2]), (2, "ab")]:
    try:
        Vector(*args)
    except (TypeError, ValueError) as exc:
        print(f"  Vector{args} -> {type(exc).__name__}: {exc}")
```

**说明：** `dim` 必须是非负整数，`elements` 必须是列表或元组，且长度等于 `dim`。只传维度时用 `[0] * dim` 生成零向量。元素在内部保存为元组，避免外部通过列表原地修改向量内容；构造时仍接受题目要求的列表或元组。

### 2. 维度与元素应为类属性还是实例属性？

**答：应为实例属性。** 每个向量对象都有自己独立的维度和元素，例如 `Vector(2, (1, 2))` 与 `Vector(3, (1, 2, 3))` 的 `dim`、`elements` 显然不同。如果在类体中直接写 `dim = ...`、`elements = ...`，它们会成为类属性，被所有实例共享；修改一个向量的元素会影响其他向量，无法表达“每个对象有自己的状态”。因此应在 `__init__` 中通过 `self.dim`、`self.elements` 绑定为实例属性。类属性适合所有实例共享的常量或默认配置，而不是每个对象各自变化的数据。

### 3. 比较运算符

两个 `Vector` 坐标相同则相等；否则按 Python 元组的字典序比较（先比第一个坐标，再比第二个，依此类推）。比较维度不同的向量时抛出异常。

```python
print("比较 Vector:")
v1 = Vector(2, (2, 4))
v2 = Vector(2, (2, 5))
print("  Vector(2,(2,4)) <  Vector(2,(2,5)):", v1 < v2)
print("  Vector(2,(2,4)) <= Vector(2,(2,4)):", v1 <= Vector(2, (2, 4)))
print("  Vector(2,(2,4)) == Vector(2,(2,4)):", v1 == Vector(2, (2, 4)))
print("  Vector(2,(2,4)) >  Vector(2,(2,3)):", v1 > Vector(2, (2, 3)))

try:
    v1 == Vector(3, (1, 2, 3))
except ValueError as exc:
    print("  维度不同时比较 -> ValueError:", exc)

try:
    v1 < Vector(3, (1, 2, 3))
except ValueError as exc:
    print("  维度不同时排序 -> ValueError:", exc)
```

**说明：** 相等比较先检查类型和维度，再比较元素元组；排序比较直接用元组的字典序，因此 `(2, 4) < (2, 5)`、`(2, 4) > (2, 3)`。维度不同时统一抛出 `ValueError`，与题目“比较维度不同的两个向量时应抛出异常”一致；与非 `Vector` 对象比较时，`==` 返回 `NotImplemented`（最终为 `False`），排序运算由 Python 抛出 `TypeError`。

### 4. `Vector.dot`

接收另一个 `Vector`，返回两向量的内积；参数类型不正确或维度不一致时抛出异常。

```python
print("Vector.dot:")
print("  Vector(3,(1,2,3)).dot(Vector(3,(4,5,6))) =",
      Vector(3, (1, 2, 3)).dot(Vector(3, (4, 5, 6))))

try:
    Vector(2, (1, 2)).dot(Vector(3, (1, 2, 3)))
except ValueError as exc:
    print("  维度不一致 -> ValueError:", exc)

try:
    Vector(2, (1, 2)).dot((1, 2))
except TypeError as exc:
    print("  参数不是 Vector -> TypeError:", exc)
```

**说明：** 内积为对应元素乘积之和，结果为 `float`。先检查参数类型，再检查维度，分别抛出 `TypeError` 和 `ValueError`。

### 5. 乘法运算符

标量左乘或右乘（`2 * v`、`v * 2`）返回各元素乘以标量的新 `Vector`；两个 `Vector` 相乘（`v * w`）返回对应位置元素相乘的新 `Vector`；维度不一致时抛出异常。

```python
print("Vector 乘法:")
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print("  v =", v, ", w =", w)
print("  2 * v =", 2 * v)
print("  v * 2 =", v * 2)
print("  v * w =", v * w)

try:
    v * Vector(3, (1, 2, 3))
except ValueError as exc:
    print("  维度不一致 -> ValueError:", exc)
```

**说明：** `__mul__` 先判断参数是否为数值：是则返回标量乘的新向量；否则判断是否为 `Vector`，维度一致时逐元素相乘。`__rmul__` 支持 `2 * v` 这种左乘写法。标量乘法不修改原向量，两个向量相乘的结果也是新对象；维度不一致时抛出 `ValueError`。
