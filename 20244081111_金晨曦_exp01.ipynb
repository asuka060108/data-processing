## 实验1：基本运算、数据类型与字符串

### 1.1 四种算术运算符：`/`、`//`、`%`、`**`

- `/`：真除法，结果**始终是 float**；
- `//`：地板除（向下取整的除法），int 之间运算得 int；
- `%`：取模（求余数）；
- `**`：幂运算。

```python
expressions = {
    "8 / 3": 8 / 3,
    "8 // 3": 8 // 3,
    "8 % 3": 8 % 3,
    "2 ** 7": 2 ** 7,
}
for expr, value in expressions.items():
    print(f"{expr:>7} = {value:<8} 类型: {type(value).__name__}")
```

### 1.2 四种基本数据类型，用 `type()` 检查

```python
int_val = 42
float_val = 3.14
str_val = "数据处理"
bool_val = True

for val in [int_val, float_val, str_val, bool_val]:
    print(f"{val!r:<12} 类型: {type(val).__name__}")
```

### 1.3 类型转换

- `int("123")`：内容为合法整数的字符串可转 int；
- `float("3.14")`：转为浮点数；
- `int(3.9)`：对浮点数**截断小数部分**（向零取整，不是四舍五入），得 3；
- `int("goat")`：字符串不是合法数字，抛出 `ValueError`。

```python
print("int(\"123\") =", int("123"))
print("float(\"3.14\") =", float("3.14"))
print("int(3.9) =", int(3.9))

try:
    int("goat")
except ValueError as e:
    print("捕获 ValueError：", e, "—— 字符串不是合法数字，无法转换")
```

### 1.4 字符串常用方法：去空白、大小写、切片、拆分、连接

```python
s = " Python Data Analysis "
print("原字符串:      ", repr(s))
print("strip 去首尾空白:", repr(s.strip()))
print("upper 全大写:   ", repr(s.upper()))
print("lower 全小写:   ", repr(s.lower()))
print("切片 [1:7]:     ", repr(s[1:7]))

words = s.split()
print("split 拆成单词: ", words)
print("空格 join 连接: ", " ".join(words))
print("短横线 join:    ", "-".join(words))
```

### 1.5 字符串拼接 / 重复与不可变性

- `+` 拼接，`*` 重复；
- 字符串是**不可变类型**，不能直接修改某个字符，否则抛出 `TypeError`；
- 只能通过切片、拼接等方式**构造一个新字符串**。

```python
print("拼接 \"a\" + \"b\":", "a" + "b")
print("重复 \"ab\" * 3:", "ab" * 3)

word = "cat"
try:
    word[0] = "b"
except TypeError as e:
    print("捕获 TypeError：", e)

new_word = "b" + word[1:]
print("构造新字符串:", new_word)
```

## 实验2：函数、条件、循环与递归

### 2.1 `score_level(score)` 成绩等级

| 分数区间 | 等级 |
|---|---|
| 90~100 | 优秀 |
| 80~不足90 | 良好 |
| 60~不足80 | 及格 |
| 0~不足60 | 不及格 |

超出 0~100 范围抛出 `ValueError`。

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


for s in [100, 95, 85, 80, 70, 60, 59, 0]:
    print(f"{s:>3} 分 -> {score_level(s)}")

for s in [-1, 101]:
    try:
        score_level(s)
    except ValueError as e:
        print(f"{s} 分 -> 捕获 ValueError：{e}")
```

### 2.2 分别用 `for` 和 `while` 计算 1~100 的整数和

```python
sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print("for   循环结果:", sum_for)
print("while 循环结果:", sum_while)
print("两者是否相等:", sum_for == sum_while)
```

### 2.3 递归版与迭代版阶乘，约定 $0! = 1$，测试 n = 0、1、5、10

```python
def factorial_rec(n):
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)


def factorial_iter(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


print(f"{'n':>3} | {'递归版':>10} | {'迭代版':>10} | 是否相等")
for n in [0, 1, 5, 10]:
    r, it = factorial_rec(n), factorial_iter(n)
    print(f"{n:>3} | {r:>10} | {it:>10} | {r == it}")
```

### 2.4 `summary(values)`：以元组返回最小值、最大值、平均值，并元组解包

```python
def summary(values):
    return min(values), max(values), sum(values) / len(values)


data = [85, 58, 92, 76, 60]
low, high, avg = summary(data)
print(f"最小值: {low}, 最大值: {high}, 平均值: {avg:.2f}")
```

### 2.5 `sum_all(*args)`：可变参数求和，测试无参 / 单参 / 多参

```python
def sum_all(*args):
    total = 0
    for x in args:
        total += x
    return total


print("无参数:  ", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4, 5))
```

## 实验3：列表、字典、元组与基础统计

使用模拟成绩数据（非真实学生信息）。

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
```

### 3.1 用 `zip()` 构建姓名→成绩字典，`items()` 逐项输出

```python
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name:<6} {score}")
```

### 3.2 用 `enumerate()` 输出序号、姓名、成绩，序号从 1 开始

```python
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{idx}. {name:<6} {score}")
```

### 3.3 列表推导式：提取及格姓名、生成成绩平方列表

```python
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("及格学生:", passed_names)
print("成绩平方:", squared_scores)
```

### 3.4 基础统计：人数、平均分、最高分、最低分、及格率

```python
count = len(scores)
mean_score = sum(scores) / count
highest, lowest = max(scores), min(scores)
pass_count = sum(1 for score in scores if score >= 60)
pass_rate = pass_count / count

print(f"人数:   {count}")
print(f"平均分: {mean_score:.2f}")
print(f"最高分: {highest}")
print(f"最低分: {lowest}")
print(f"及格率: {pass_rate:.0%}")
```

### 3.5 别名 vs 浅拷贝：`alias = scores` 与 `copied = scores.copy()`

为避免改动后续实验使用的原始成绩，这里使用**单独的演示列表**。

```python
demo = [1, 2, 3]
alias = demo
copied = demo.copy()

alias[0] = 99
print("原列表 demo:", demo)
print("别名 alias:", alias)
print("副本 copied:", copied)
print("demo is alias:", demo is alias, "| demo is copied:", demo is copied)
```

**结论：**
- 直接赋值 `alias = demo` 只是给同一列表起别名，修改 `alias` 会影响原列表；
- `copy()` 得到新列表，修改副本不影响原列表（浅拷贝）。

### 3.6 元组不可变；不等长序列的 `zip()`

```python
record = ("Alice", 85)
try:
    record[1] = 90
except TypeError as e:
    print("捕获 TypeError：", e)

print("不等长 zip([1,2,3], ['a','b']):", list(zip([1, 2, 3], ["a", "b"])))
print("结论：zip 以较短的序列为准，多出的元素被丢弃。")
```

## 实验4：文件读写、异常处理与模块

### 4.1 查看工作目录、构造路径并创建数据目录

```python
import os

work_dir = os.getcwd()
print("当前工作目录:", work_dir)
data_dir = os.path.join(work_dir, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录:", data_dir)
```

### 4.2 写入 `students.txt`（UTF-8），再读取并用字符串方法解析

```python
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("跳过表头后的解析结果：")
for line in lines[1:]:
    print(line.strip().split(","))
```

### 4.3 `parse_student(line)` 函数

校验：字段数量必须为 2、姓名不能为空、成绩必须是 0~100 的整数，否则抛 `ValueError`。

```python
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量不是2: {line.strip()!r}")
    name, score_str = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是合法整数: {score_str!r}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return {"name": name, "score": score}
```

### 4.4 逐行处理，捕获无效记录并记录行号与原因，继续处理后续行

```python
valid_records = []
invalid_records = []

with open(txt_path, "r", encoding="utf-8") as f:
    all_lines = f.readlines()

for lineno, line in enumerate(all_lines[1:], start=2):
    if not line.strip():
        continue
    try:
        valid_records.append(parse_student(line))
    except ValueError as e:
        invalid_records.append((lineno, line.strip(), str(e)))
        print(f"第 {lineno} 行无效: {line.strip()} -> 原因: {e}")

print("\n有效记录:", valid_records)
print("无效记录数:", len(invalid_records))
```

### 4.5 统计结果写入 `report.txt`，分别演示 `w` 与 `a` 模式

```python
n_valid = len(valid_records)
n_invalid = len(invalid_records)
mean_v = sum(r["score"] for r in valid_records) / n_valid
pass_rate_v = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

report_path = os.path.join(data_dir, "report.txt")

# w 模式：打开时清空文件，覆盖已有内容
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {n_valid}\n")
    f.write(f"无效记录数: {n_invalid}\n")
    f.write(f"平均分: {mean_v:.2f}\n")
    f.write(f"及格率: {pass_rate_v:.0%}\n")

# a 模式：在文件末尾追加，保留已有内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("（a 模式追加）报告生成完毕\n")

with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
```

**说明：** `w` 模式每次打开都会清空原内容（覆盖写）；`a` 模式把内容追加到文件末尾，原有内容保留。

### 4.6 用 `pickle` 保存并读取，验证恢复数据与原数据相等

## 实验5：类、继承与对象复制

### 5.1 定义 `Student` 类：`__init__`、`is_passed()`、`__str__()`

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"学生 {self.name}，成绩 {self.score}"
```

### 5.2 为实验4的每条有效记录创建 `Student` 对象

```python
students = [Student(r["name"], r["score"]) for r in valid_records]
for stu in students:
    print(f"{stu} | 及格状态: {stu.is_passed()}")
```

### 5.3 `GraduateStudent` 子类：增加 `research_topic`，用 `super()` 初始化，重写 `__str__()`

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"研究生 {self.name}，成绩 {self.score}，研究方向: {self.research_topic}"


grad = GraduateStudent("Alice", 85, "数据挖掘")
print(grad)
print("及格状态:", grad.is_passed())
```

### 5.4 直接赋值 / `copy.copy()` / `copy.deepcopy()` 对比

对含嵌套列表的对象进行三种复制，修改内部列表。**每次比较都从新建原始对象开始。**

```python
import copy

def make_original():
    return {"name": "Alice", "scores": [85, 90]}

# 1) 直接赋值：两个引用指向同一个对象
original = make_original()
alias_obj = original
alias_obj["scores"].append(95)
print("【直接赋值】")
print("  original:", original)
print("  alias   :", alias_obj)
print("  是同一对象:", original is alias_obj)

# 2) 浅拷贝：外层是新对象，内部嵌套列表仍共享
original = make_original()
shallow_obj = copy.copy(original)
shallow_obj["scores"].append(95)
print("【浅拷贝 copy.copy】")
print("  original:", original)
print("  shallow :", shallow_obj)
print("  外层同一对象:", original is shallow_obj, "| 内部列表共享:", original["scores"] is shallow_obj["scores"])

# 3) 深拷贝：所有层级都独立复制
original = make_original()
deep_obj = copy.deepcopy(original)
deep_obj["scores"].append(95)
print("【深拷贝 copy.deepcopy】")
print("  original:", original)
print("  deep    :", deep_obj)
print("  内部列表共享:", original["scores"] is deep_obj["scores"])
```

**结论：**
- **直接赋值**：不产生新对象，两个引用完全共享；
- **浅拷贝 `copy.copy()`**：产生新的外层对象，但内层嵌套对象仍是共享的，修改内层列表会互相影响；
- **深拷贝 `copy.deepcopy()`**：递归复制所有层级，原对象与副本完全独立，互不影响。
